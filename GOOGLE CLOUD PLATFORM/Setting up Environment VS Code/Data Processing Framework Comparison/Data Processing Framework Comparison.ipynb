```python
#using pandas
import glob
import pandas as pd

files= glob.glob('data/nyse_all/nyse_data/*')

rec_count=0

for file in files:
    df=pd.read_csv(
        file,
        names=['stock_id','trans_date','open_price','low_price','high_price','volume']
               )
    rec_count+=df.shape[0]

rec_count
```

```python
#using dask
#dask uses full capacity of the system
#uses all the servers effectively

import dask.dataframe as dd
df=dd.read_csv(
    'data/nyse_all/nyse_data/*',
    names=['stock_id','trans_date','open_price','low_price','high_price','volume']
)

df.shape[0].compute()
```

```python
#using pyspark

from pyspark.sql import SparkSession

spark=SparkSession.builder.appName('NYSE count').master('local').getOrCreate()


df=spark.read.csv('data/nyse_all/nyse_data/*',
                  schema=''' 
                  stock_id STRING,trans_date STRING, open_price float, low_price float,
                   highr_price float,close_price float, volume bigint''' )

df.count()
```
